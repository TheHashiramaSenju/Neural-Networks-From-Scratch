```python
#for logistic regression we use log loss
import keras
import numpy as np # Import numpy

# Define X_train_flattened and y_train here before using them.
# (Example placeholders - replace with your actual data loading and preprocessing)
X_train_flattened = np.random.rand(100, 784) # Example: 100 samples, 784 features
y_train = np.random.randint(0, 2, 100)      # Example: 100 binary labels

model = keras.Sequential([
    keras.layers.Dense(1, input_shape=(784,), activation='sigmoid')
])

print("BCE")

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X_train_flattened, y_train, epochs=50)

print("MSE")

model.compile(optimizer='adam',
              loss='mean_squared_error',
              metrics=['accuracy'])

model.fit(X_train_flattened, y_train, epochs=50)

print("MAE")
model.compile(optimizer='adam',
              loss='mean_absolute_error',
              metrics=['accuracy'])

model.fit(X_train_flattened, y_train, epochs=50)
```

```python
import numpy as np
y_predicted = np.array([1,1,0,0,1])
y_true = np. array([0.30,0.7,1,0,0.5])
def mae(y_true, _predicted) :
  total_error = 0
  for yt, yp in zip(y_true, _predicted):
    total_error += abs ( yt - yp)
  print ("Total error:", total_error)
  mae = total_error / len(y_true)
  return mae
mae(y_true, y_predicted)
```
