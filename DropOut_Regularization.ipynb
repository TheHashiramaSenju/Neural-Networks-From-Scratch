```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv("/content/sonar_dataset.csv", header=None)
df.sample(5)
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
df[60].value_counts()
```

```python
#we drop the final column for training purposes
X = df.drop(60, axis=1)
y = df[60]
y.head()
```

```python
y = pd.get_dummies(y, drop_first=True)
y.sample(5)
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
X_train.shape, X_test.shape
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
def get_model():
  model = keras.Sequential([
      keras.layers.Dense(60, input_dim=60, activation='relu'),
      keras.layers.Dense(30, activation='relu'),
      keras.layers.Dense(15, activation='relu'),
      keras.layers.Dense(1, activation='sigmoid') #BINARY CLASSIFICATION PROBLEM

    ])
  model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
  model.fit(X_train, y_train, epochs=100, batch_size=8)
  return model

```

```python
with tf.device('/device:GPU:0'):
  model = get_model()
  model.evaluate(X_test, y_test)
```

```python
model.evaluate(X_test, y_test)
```

```python
y_pred = model.predict(X_test).reshape(-1)
print(y_pred[:10])

#round the values to the nearest integer like in 0 and 1 range

y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(classification_report(y_test, y_pred))
```

```python
modeld = keras.Sequential([
      keras.layers.Dense(60, input_dim=60, activation='relu'),
      keras.layers.Dropout(0.5), #drops 50 percent of the neurons present here
      keras.layers.Dense(30, activation='relu'),
      keras.layers.Dropout(0.5),
      keras.layers.Dense(15, activation='relu'),
      keras.layers.Dropout(0.5),
      keras.layers.Dense(1, activation='sigmoid') #BINARY CLASSIFICATION PROBLEM

    ])
modeld.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
modeld.fit(X_train, y_train, epochs=100, batch_size=8)
```

Using dropout will reduce the accuracy on the training set and we care about the test set here. We must know why and where here(Know it here)

```python
modeld.evaluate(X_test, y_test)
```

```python
y_pred = modeld.predict(X_test).reshape(-1)
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

```python
print(classification_report(y_test, y_pred))
```
